#Cleaning Dataset

##Strip Whitespace

```python
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].str.strip()
```

## 3. Fix Mismatched Values
3 rows had duration (e.g. `74 min`) stored in the `rating` column and `NaN` in `duration`.

```python
mask = df["rating"].str.endswith(" min", na=False) & df["rating"].str.split(" ").str[0].str.isnumeric()
df.loc[mask, "duration"] = df.loc[mask, "rating"]
df.loc[mask, "rating"]   = np.nan
print(f"[FIX]   {mask.sum()} row(s) had duration stored in 'rating' — corrected")
```

## 4. Drop Unrecoverable Rows

```python
before = len(df)
df.dropna(subset=["date_added", "duration"], inplace=True)
print(f"[DROP]  {before - len(df)} row(s) dropped (missing date_added or duration)")
```

## 5. Fill Missing Values

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"]     = df["cast"].fillna("Unknown")
df["country"]  = df["country"].fillna("Unknown")
df["rating"]   = df["rating"].fillna("NR").replace("UR", "NR")

print("Missing values after fill:")
print(df.isnull().sum())
```

## 6. Parse `date_added` → datetime + helper columns

```python
df["date_added"]  = pd.to_datetime(df["date_added"].str.strip(), format="%B %d, %Y")
df["year_added"]  = df["date_added"].dt.year
df["month_added"] = df["date_added"].dt.month

df[["date_added", "year_added", "month_added"]].head()
```

## 7. Extract Duration Value & Unit

```python
split = df["duration"].str.split(" ", n=1, expand=True)
df["_dur_val"]  = pd.to_numeric(split[0], errors="coerce")
df["_dur_unit"] = split[1].str.lower().map(
    lambda x: "minutes" if x == "min" else "seasons" if "season" in str(x) else np.nan
)
df[["duration", "_dur_val", "_dur_unit"]].head()
```

## 8. Convert Movie Duration (minutes) → HH:MM:SS

```python
def mins_to_hhmmss(mins):
    h, m = divmod(int(mins), 60)
    return f"{h:02d}:{m:02d}:00"

df["duration"] = np.where(
    df["_dur_unit"] == "minutes",
    df["_dur_val"].apply(lambda v: mins_to_hhmmss(v) if pd.notna(v) else np.nan),
    np.nan
)

df[df["type"] == "Movie"][["title", "duration"]].head()
```

## 9. Add `seasons` Column (TV Shows only)

```python
df["seasons"] = np.where(
    df["_dur_unit"] == "seasons",
    df["_dur_val"].astype("Int64"),
    pd.NA
)

df[df["type"] == "TV Show"][["title", "duration", "seasons"]].head()
```

## 10. Add Helper Columns

```python
df["is_movie"]   = (df["type"] == "Movie").astype(bool)
df["num_genres"] = df["listed_in"].str.split(",").str.len()
df["num_cast"]   = df["cast"].apply(
    lambda x: len(x.split(",")) if x != "Unknown" else 0
)
```

## 11. Drop Temp Columns & Set Final Column Order

```python
df.drop(columns=["_dur_val", "_dur_unit"], inplace=True)

df = df[[
    "show_id", "type", "is_movie", "title",
    "director", "cast", "num_cast",
    "country", "date_added", "year_added", "month_added",
    "release_year", "rating",
    "duration", "seasons",
    "listed_in", "num_genres", "description",
]]

df.head(10)
```

```python
df.to_csv('cleaned_dataset.csv', index=False)
print("SAVED")
```

```python
def check_movie_access():
    # 1. Age Mapping Dictionary (Standard Guidelines)
    rating_age_map = {
        'TV-Y': 0, 'TV-G': 0, 'G': 0,
        'TV-Y7': 7, 'TV-Y7-FV': 7,
        'TV-PG': 7, 'PG': 7,
        'PG-13': 13,
        'TV-14': 14,
        'TV-MA': 17, 'R': 17, 'NC-17': 18, 'NR': 18
    }

    # 2. Get User Input
    movie_input = input("Enter the movie or show title: ").strip()

    # Search for the title (case-insensitive)
    match = df[df['title'].str.lower() == movie_input.lower()]

    if match.empty:
        print(f"Sorry, '{movie_input}' was not found in the database.")
        return

    # Get the rating of the first match
    movie_rating = match.iloc[0]['rating']
    min_age = rating_age_map.get(movie_rating, 0) # Default to 0 if rating unknown

    try:
        user_age = int(input("Enter your age: "))
    except ValueError:
        print("Invalid age entered. Please enter a number.")
        return

    # 3. Logic for Access
    print("-" * 30)
    print(f"Title: {match.iloc[0]['title']}")
    print(f"Rating: {movie_rating} (Minimum suggested age: {min_age})")

    if user_age >= min_age:
        print("RESULT: ACCESS GRANTED. Enjoy the show!")
    else:
        print(f"RESULT: ACCESS DENIED. You must be at least {min_age} years old.")
    print("-" * 30)

# Run the program
check_movie_access()
```
